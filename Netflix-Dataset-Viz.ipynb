# Netflix Data set Visualization Project


* Import the Libraries

```python
import pandas as pd
from IPython.core.pylabtools import figsize
from matplotlib import pyplot as plt
```

* Counting the Movies and TV

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])

# Count movies vs TV shows
type_count = df['type'].value_counts()

# Create the bar chart
plt.figure(figsize=(6, 4))
plt.bar(type_count.index, type_count.values, color=['skyblue', 'yellow'])

plt.title('Netflix Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')

plt.tight_layout()
plt.savefig('netflix_movies_vs_tv.png')
plt.show()



```

* Counting the Rating

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])

# Count the ratings
rating_count = df['rating'].value_counts()

# Create the pie chart
plt.figure(figsize=(8, 6))
plt.pie(rating_count, labels=rating_count.index , autopct = '%1.1f%%' , startangle = 90)

plt.title('Netflix Content Ratings ')

plt.tight_layout()
plt.savefig('netflix_rating_pie.png')
plt.show()


```

* Distribution of Movies Duration

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])

movie_df = df[df['type'] == 'Movie'].copy()
movie_df['duration_int'] = movie_df['duration'].str.replace('min', '').astype(int)


# Create the histogram
plt.figure(figsize=(8, 6))
plt.hist(movie_df['duration_int'], bins = 30  , color = 'purple' , edgecolor='black' )

plt.title('Distribution of Movies Duration')
plt.xlabel('Duration')
plt.ylabel('Movies')

plt.tight_layout()
plt.savefig('netflix_movies_count_histogram.png')
plt.show()


```

* Release year VS No of Shows

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])


release_count = df['release_year'].value_counts().sort_index()

# Create the Scatter Chart
plt.figure(figsize=(8,6))
plt.scatter(release_count.index , release_count.values,color = 'red')

plt.title('Release Year VS Movies')
plt.xlabel('Release Year')
plt.ylabel('Movies')

plt.tight_layout()
plt.savefig('netflix_movies_release_scatter.png')
plt.show()
```

* Top 10 Countries by No of Shows

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])

# Create the Horizontal Bar  Chart
country_count = df['country'].value_counts().head(10)
plt.figure(figsize=[8 ,6])
plt.barh(country_count.index , country_count.values, color = 'teal')


plt.title('Top 10 Countries by No of Shows')
plt.xlabel('No of Shows')
plt.ylabel('Countries')

plt.tight_layout()
plt.savefig('netflix_top_coutries.barh.png')
plt.show()

```

* Release Year of Movies and TV Shows

```python
# loading the data
df = pd.read_csv("netflix_titles.csv")

# Cleaning the data by removing NaN values
df = df.dropna(subset=['type', 'release_year', 'rating', 'country', 'duration'])

# Create the Plot Lines

content_by_year = df.groupby(['release_year','type']).size().unstack().fillna(0)
fig, ax = plt.subplots(1,2 , figsize = (12,5))

# First subplot : Movies
ax[0].plot(content_by_year.index, content_by_year['Movie'] , color = 'blue')
ax[0].set_title('Movies released by Year')
ax[0].set_xlabel('Year')
ax[0].set_ylabel('No of Movies')

# First subplot : Shows
ax[1].plot(content_by_year.index, content_by_year['TV Show'] , color = 'orange')
ax[1].set_title('Shows released by Year')
ax[1].set_xlabel('Year')
ax[1].set_ylabel('No of Shows')

fig.suptitle('Comparison of Movies and TV Shows over Year')
plt.tight_layout()
plt.savefig('netflix_movies-shows_compare.png')
plt.show()
```
